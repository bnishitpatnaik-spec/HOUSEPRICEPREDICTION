```python
import pandas as pd

df = pd.read_csv('housing_data.csv')

furnishing_mapping = {'unfurnished': 0, 'semi-furnished': 1, 'furnished': 2}

if df['furnishingstatus'].dtype == 'O':
    df['furnishingstatus'] = df['furnishingstatus'].str.strip().str.lower().map(furnishing_mapping)
    print("Encoding applied successfully!")
else:
    print("Column is already numerical. No changes needed.")

if df['furnishingstatus'].isnull().any():
    print("Warning: Some values didn't match the mapping. Filling NaNs with 0.")
    df['furnishingstatus'] = df['furnishingstatus'].fillna(0).astype(int)

df.to_csv('housing_data_cleaned.csv', index=False)
```

```python
df
```

```python
binary_mapping = {'yes': 1, 'no': 0}
binary_cols = ['mainroad', 'guestroom', 'basement', 'airconditioning']

for col in binary_cols:
    df[col] = df[col].astype(str).str.strip().str.lower().map(binary_mapping)

df.to_csv('housing_data_final.csv', index=False)
```

```python
df.drop('prefarea',axis=1,inplace=True)
df.drop('hotwaterheating',axis=1,inplace=True)
df.to_csv('housing_data_final.csv', index=False)
```

```python
price_mean=df['price'].mean()
price_std=df['price'].std()
df['price_scaled']=(df['price']-price_mean)/price_std;
df.to_csv('housing_data_final.csv', index=False)
```

```python
area_mean=df['area'].mean()
area_std=df['area'].std()
df['area_scaled']=(df['area']-area_mean)/area_std;
df.to_csv('housing_data_final.csv', index=False)
```
